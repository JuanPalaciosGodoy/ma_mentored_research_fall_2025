# How to find routes inside a market?

```python
routes = pl.DataFrame(queries.get_routes_in_market(market_id='natural-gas')['response']['routes'])
routes.style
```

# How to find subroutes inside a market?

sometimes markets don't have subroutes. so you can simply pull data using the market_id and the route_id (no need for a subroute)

```python
subroutes = pl.DataFrame(queries.get_subroutes_in_market(market_id='natural-gas', route_id='pri')['response']['routes'])
subroutes.style
```

# How to get data from EIA?

```python
response = queries.get_eia_data(
    market_id='natural-gas',
    route_id='pri',
    subroute_id='fut',
    data=['value'],
    frequency="daily",
    start_date="2025-01-01"
)
data = pl.DataFrame(response['response']['data'])
data
```

```python
# front month future contract
front_month_contract = data.sort(by="period")

# Create the line plot
fig = px.line(front_month_contract, x="period", y="value")

# Display the plot
fig.show()
```
