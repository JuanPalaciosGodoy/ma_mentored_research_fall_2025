```python
import queries
from enums import METADATA, Route, MarketUrl, Markets
import polars as pl
```

```python
market_url = Markets.ELECTRICITY.value
route = Route.STATE_ELECTRICITY_PROFILES.value
data_type = MarketUrl.EMISSIONS_BY_STATE_DATE.value

metadata = METADATA[market_url][route][data_type]
metadata
```

```python
data = [value for key, value in metadata['data'].items()]
frequency = metadata["frequency"][1]
```

```python
json_response = queries.get_eia_electricity_query(
    route=route,
    data_type=data_type,
    market_url=market_url,
    data=data,
    frequency=frequency
)
```
